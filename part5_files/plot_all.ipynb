Use this code to generate a plot of all cases of `c_roi` saved in folders within this directory.

```python
import os
import numpy as np
from matplotlib import pyplot as plt
test_folders = os.listdir(os.getcwd())
for i in range(len(test_folders)):
    if os.path.isdir(test_folders[i]):
        if os.path.isfile(f"{test_folders[i]}/c_roi.txt"):
            cur_data = np.loadtxt(f"{test_folders[i]}/c_roi.txt")
            plt.plot(cur_data[:,0], cur_data[:,1],label=test_folders[i])
plt.xlabel('Time (s)')
plt.ylabel('Concentration (molecules/sq micron)')
plt.legend()
plt.show()
```
